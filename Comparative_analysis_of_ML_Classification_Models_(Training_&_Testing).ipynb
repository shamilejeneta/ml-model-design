```python
from google.colab import drive
import os

# Connect Google Drive
drive.mount('/content/drive')

# Names of the required files
train_filename = "TRAIN_AI4I2020_PDM_DATASET.xlsx"
test_filename = "TEST_AI4I2020_PDM_DATASET.xlsx"

# Search the entire MyDrive for the files
train_file = None
test_file = None

for root, dirs, files_in_folder in os.walk('/content/drive/MyDrive'):
    if train_filename in files_in_folder:
        train_file = os.path.join(root, train_filename)

    if test_filename in files_in_folder:
        test_file = os.path.join(root, test_filename)

# Check whether files were found
if train_file is None:
    print("ERROR: Training file was not found in Google Drive.")
else:
    print("Training file found:")
    print(train_file)

if test_file is None:
    print("ERROR: Testing file was not found in Google Drive.")
else:
    print("Testing file found:")
    print(test_file)
```

```python
import pandas as pd

# Check that both files were found
if train_file is None or test_file is None:
    raise FileNotFoundError(
        "One or both Excel files could not be found in Google Drive. "
        "Please check the filenames."
    )

# Read the Excel files
train_df = pd.read_excel(train_file)
test_df = pd.read_excel(test_file)

print("Training DataFrame created successfully!")
print("Testing DataFrame created successfully!")

print("\nTraining Dataset Shape:")
print(train_df.shape)

print("\nTesting Dataset Shape:")
print(test_df.shape)

print("\nTraining Dataset:")
display(train_df.head())

print("\nTesting Dataset:")
display(test_df.head())
```

```python
# ==========================================================
# TRAINING DATA
# ==========================================================

# First five columns A-E = Features
X_train = train_df.iloc[:, 0:5]

# Last column F = Label
y_train = train_df.iloc[:, 5]


# ==========================================================
# TESTING DATA
# ==========================================================

# First five columns A-E = Features
X_test = test_df.iloc[:, 0:5]

# Last column F = Label
y_test = test_df.iloc[:, 5]


# Display the results
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

print("\nTraining Features (X_train):")
display(X_train.head())

print("\nTraining Labels (y_train):")
display(y_train.head())

print("\nTesting Features (X_test):")
display(X_test.head())

print("\nTesting Labels (y_test):")
display(y_test.head())
```

```python
# ==========================================================
# STEP 5: CREATE AND TRAIN 3 ML MODELS
# ==========================================================

from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

# ==========================================================
# 1. RANDOM FOREST
# ==========================================================

random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train Random Forest using training data
random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")


# ==========================================================
# 2. DECISION TREE
# ==========================================================

decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

# Train Decision Tree using training data
decision_tree_model.fit(X_train, y_train)

print("Decision Tree model trained successfully!")


# ==========================================================
# 3. SUPPORT VECTOR MACHINE (SVM)
# ==========================================================

svm_model = SVC(
    kernel='rbf',
    random_state=42
)

# Train SVM using training data
svm_model.fit(X_train, y_train)

print("SVM model trained successfully!")
```

```python
# ==========================================================
# STEP 6: PREDICT USING X_TEST
# ==========================================================

# Random Forest prediction
y_pred_rf = random_forest_model.predict(X_test)

# Decision Tree prediction
y_pred_dt = decision_tree_model.predict(X_test)

# SVM prediction
y_pred_svm = svm_model.predict(X_test)


print("Predictions generated successfully!")

print("\nRandom Forest predictions:")
print(y_pred_rf[:10])

print("\nDecision Tree predictions:")
print(y_pred_dt[:10])

print("\nSVM predictions:")
print(y_pred_svm[:10])
```

```python
# ==========================================================
# STEP 7: EVALUATION METRICS
# ==========================================================

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


def evaluate_model(model_name, y_test, y_pred):

    print("\n")
    print("=" * 60)
    print(model_name)
    print("=" * 60)

    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)

    print("\nConfusion Matrix:")
    print(cm)

    # Accuracy
    accuracy = accuracy_score(y_test, y_pred)

    # Precision
    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    # Recall
    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    # F1 Score
    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )

    print("\nAccuracy :", accuracy)
    print("Precision:", precision)
    print("Recall   :", recall)
    print("F1-Score :", f1)

    return accuracy, precision, recall, f1


# ==========================================================
# RANDOM FOREST EVALUATION
# ==========================================================

rf_results = evaluate_model(
    "RANDOM FOREST",
    y_test,
    y_pred_rf
)


# ==========================================================
# DECISION TREE EVALUATION
# ==========================================================

dt_results = evaluate_model(
    "DECISION TREE",
    y_test,
    y_pred_dt
)


# ==========================================================
# SVM EVALUATION
# ==========================================================

svm_results = evaluate_model(
    "SUPPORT VECTOR MACHINE (SVM)",
    y_test,
    y_pred_svm
)
```

```python
# ==========================================================
# STEP 9: SAVE TRAINED MODELS AS PKL FILES
# ==========================================================

import pickle

# Save Random Forest
with open("random_forest_model.pkl", "wb") as file:
    pickle.dump(random_forest_model, file)

# Save Decision Tree
with open("decision_tree_model.pkl", "wb") as file:
    pickle.dump(decision_tree_model, file)

# Save SVM
with open("svm_model.pkl", "wb") as file:
    pickle.dump(svm_model, file)

print("All three models saved successfully!")

print("\nFiles created:")
print("1. random_forest_model.pkl")
print("2. decision_tree_model.pkl")
print("3. svm_model.pkl")
```
